### Median value of Horsepower (before and after filling missing values)

```python
median_before=df.horsepower.median()
print('Median before:', median_before)
```

```python
mode=df.horsepower.mode()[0]
print('Mode:', mode)
```

```python
df['horsepower']=df['horsepower'].fillna(mode)
```

```python
median_after=df.horsepower.median()
print('Median after:', median_after)
```

```python
if median_after > median_before:
    print('Yes, it is increased')
elif median_after<median_before:
    print('NO, it has not increased')
else:
    print('Either Both are Equal or there must be something wrong in the dataset')
```

### Sum of weights

```python
df[df['origin']=='Asia']
```

```python
df[df['origin']=='Asia'][['vehicle_weight', 'model_year']]
```

```python
df[df['origin']=='Asia'][['vehicle_weight', 'model_year']].head(n=7)
```

```python
X=df[df['origin']=='Asia'][['vehicle_weight', 'model_year']].head(n=7)
X
```

```python
XTX=X.T.dot(X)
XTX
```

```python
XTX_inv=np.linalg.inv(XTX)
XTX_inv
```

```python
y=np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv.dot(X.T).dot(y)
w
```

```python
w.sum()
```

```python
XTX=X.T @ X
XTX_inv=np.linalg.inv(XTX)
w=XTX_inv @ X.T @ y
w.sum().round(3)
```
