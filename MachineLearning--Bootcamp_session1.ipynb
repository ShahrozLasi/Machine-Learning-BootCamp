```python
print(np.dot(a,b))
## \sum u_i v_i
def vec_vec_multiplication(u, v):
    assert u.shape[0]==v.shape[0]
    n=u.shape[0]
    result=0.0
    for i in range(n):
        result=result+u[i]*v[i]
    return result
```

```python
print(vec_vec_multiplication(a,b))
```

```python
a=np.array([[2,4,5,6],[1,2,1,2],[3,1,2,1]])
# v=np.array([1,0.5,2,1])
```

```python
def matrix_vector_multiplication(u,v):
    assert u.shape[1]==v.shape[0]
    
    num_rows=u.shape[0]
    
    result=np.zeros(num_rows)
    
    for i in range(num_rows):
        
        result[i]=vec_vec_multiplication(u[i], v)
   
    return result
        
```

```python
print(matrix_vector_multiplication(a,b))
# print(np.matmul(a,v))
```

```python
v=np.array([[1,0,0,2], [1,0.5,2,1], [2,1,1,0]])
print(v.shape)
print(v.T.shape)
# print(np.delete(v, [0], axis=1), np)
# print(np.delete(v, [0], axis=1))
# print(v.shape)
```

```python
def matrix_matrix_multiplication(u,v):
        assert u.shape[1]==v.shape[0]
        
        num_rows=u.shape[0]
        num_cols=v.shape[1]
        
        result=np.zeros((num_rows, num_cols))
        
        for i in range(num_cols):
            vi    = v[:,i]
            uvi   = matrix_vector_multiplication(u,vi)
            result[:,i]=uvi
            
        return result
```

```python
print(matrix_matrix_multiplication(a,v.T))
print(a.shape)
print(v.T.shape)
```
